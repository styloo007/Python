```python
# Pandas

import pandas as pd

ages2 = pd.Series([22,23,34,26])
print(ages2.mean())


data = {
    'Name':['Alice','Bob', 'Shashank'],
    'Age':[25,30,22],
    'Nation':["USA", "UK", "India"]
}

df = pd.DataFrame(data)
print(df)
print(df['Age'].mean()) 
print(df['Age'].median())

ageBelow26DF = df[df['Age']<26]

print(ageBelow26DF)


#UseCase Example

marksData = {
    'Name':['A','B','C'],
    'Score':[70,55,32],
    'Passed':[True,True, False]
}
marksData = pd.DataFrame(marksData)
newRow = {'Name':'D', 'Score':88, 'Passed':True}
newRow = pd.DataFrame([newRow])
marksData =  pd.concat([marksData, newRow], ignore_index=True)


passedTrue = marksData[marksData['Passed']==True]

print(passedTrue)

averageScore = marksData['Score'].mean()

averageScore
```
