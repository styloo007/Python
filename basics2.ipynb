```python
# OOP


# Class, Attribute and Methods
class Car:
    def __init__(self,brand,color,speed):
        self.brand = brand
        self.color = color
        self.speed = speed
        
    def accelerate(self):
        self.speed +=10
        return self.speed
    

car1= Car("Toyota","Green",110)
print(car1.accelerate())

# Inheritance

class ElectricCar(Car):
    def __init__(self, brand, color, speed, battery):
        super().__init__(brand, color, speed)
        self.battery = battery
        

car2 = ElectricCar("Tesla","Black",50,"5000MaH")
print(car2.accelerate())


# Encapsulation

class Account:
    def __init__(self,balance):
        self._balance = balance
    def deposit(self,amount):
        self._balance +=amount
        print(f"Amount Deposited: {amount}")
        print(f"Total Balance: {self._balance}")
        
acc1 = Account(5000)
acc1.deposit(500)
```

```python
def find_max(list1):
    return max(list1)


def find_max2(list1):
    maxNum=list1[0]
    for i in range(1,len(list1)):
        if list1[i]>maxNum:
            maxNum=list1[i]
    return maxNum

print(find_max([1,2,3,4,5,8]))
print(find_max2([1,2,3,4,5,8]))

```

```python
def count_vowels(string1):
    count=0
    for i in range(0,len(string1)):
        if string1[i]=="a" or string1[i]=="e" or string1[i]=="i" or string1[i]=="o" or string1[i]=="u":
            count+=1
    return count

def count_vowels2(string1):
    vowels = "aeiouAEIOU"
    return sum(1 for char in string1 if char in vowels)

print(count_vowels("hello world"))
print(count_vowels2("hello world"))
```

```python
def two_sum(nums, target):
    for i in range(0,len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j] == target:
                return [i,j]
            

    return [-1,-1]




def two_sum2(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff],i]
        seen[num] = i


arr = [2, 7, 11, 15]
target = 26

print(two_sum(arr,target))
print(two_sum2(arr,target))
```

```python
def duplicates(nums):
    duplicates = []
    for i in range(0,len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i]==nums[j] and nums[i] not in duplicates:
                duplicates.append(nums[i])
    
    return duplicates


def duplicates2(nums):
    duplicates = {}
    
    for num in nums:
        duplicates[num] = duplicates.get(num,0)+1
        
    
    return duplicates



print(duplicates([1,2,3,5,4,4,5,6]))
print(duplicates2([1,2,3,5,4,4,5,6]))
                
```

```python
# Sort a List

def sortList(nums):
    nums.sort()
    return nums

def sortList2(nums):
    for i in range(len(nums)):
        already_sorted = True
        
        for j in range(len(nums)-i-1):
            if nums[j]>nums[j+1]:
                nums[j],nums[j+1]=nums[j+1],nums[j]
                already_sorted = False
        if already_sorted:
            break
    return nums
        

print(sortList([0,6,2,3,1]))
print(sortList2([0,6,2,3,1]))
```

```python
from collections import Counter

def word_frequency(string1):
    words = string1.lower().split()
    return dict(Counter(words))


def word_frequency2(string1):
    words = string1.lower().split()
    duplicates = {}
    
    for word in words:
        duplicates[word] =  duplicates.get(word,0)+1
    
    return duplicates

print(word_frequency("the quick brown fox jumps over the lazy dog the dog slept"))
print(word_frequency2("the quick brown fox jumps over the lazy dog the dog slept"))
```

```python
import string
alphabets = string.ascii_lowercase

def isPangram(text):
    text = text.lower()
    for i in alphabets:
        if i  not in text:
            return "No"
    
    
    return "Yes"

print(isPangram("The quick brown fox jumps over the lazy dog"))
```
